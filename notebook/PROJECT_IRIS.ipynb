```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset from a reliable public URL to prevent FileNotFoundError
url = "/content/iris_data.csv"
# Provide explicit column names because the CSV lacks a header row
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df = pd.read_csv(url, names=column_names)

# Display dataset
print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# 1. Count records for each species
print("\nSpecies Counts:")
print(df['species'].value_counts())
```

```python
# 2. Mean, Median, Minimum and Maximum
sepal = df['sepal_length'].to_numpy()
petal = df['petal_length'].to_numpy()

print("Sepal Length Statistics:")
print("Mean:", np.mean(sepal))
print("Median:", np.median(sepal))
print("Minimum:", np.min(sepal))
print("Maximum:", np.max(sepal))

print("\nPetal Length Statistics:")
print("Mean:", np.mean(petal))
print("Median:", np.median(petal))
print("Minimum:", np.min(petal))
print("Maximum:", np.max(petal))
```

```python
# 3. Average petal measurements by species
print("\nAverage Petal Measurements:")
print(df.groupby('species')[['petal_length', 'petal_width']].mean())
```

```python
# 4. Five flowers with largest petal length
print("\nTop 5 Flowers:")
print(df.sort_values(by='petal_length', ascending=False).head(5))
```

```python
# 5. Check missing values and duplicates
print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# 6. Compare average measurements across species
print("\n Average Measurements by Species:")
print(df.groupby('species')[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].mean())
```

```python
# 7. Species Count Bar Chart
df['species'].value_counts().plot(kind='bar', title='Count of Iris Species', xlabel='Species', ylabel='Number of Flowers')
```

```python
# 8. Petal Length Histogram
plt.hist(df['petal_length'], bins=15, edgecolor='black')

plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
